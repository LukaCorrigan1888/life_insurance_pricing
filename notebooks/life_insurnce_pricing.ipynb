```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from life_pricing.mortality import *
```

```python
# Mortality assumption configuration

config = GompertzConfig(0.0005, 1.08)
age = 50
term = 20
```

```python
# Calculate survival probabilities and death probabilities for ages 50 to 70 for exporting to Excel etc.

t = np.arange(0, term)

tpx = survival_prob(age, t, config)
pxt = survival_prob(age + t, 1, config)
qxt = 1 - survival_prob(age + t, 1, config)

df = pd.DataFrame({
    "t": t,
    "age": age + t,
    "tpx": tpx,
    "px": pxt,
    "qx": qxt,
    "tdiff_qx": tpx * qxt
})
```

```python
# Future lifetime distribution for a life aged 50

t = np.linspace(0, 40, 1000)
ages = age + t

pdf = mu(ages, config) * survival_prob(age, t, config)

plt.figure()
plt.plot(ages, pdf)
plt.xlabel("Age")
plt.ylabel("f_T50(t)")
plt.title("PDF of future lifetime for a life aged 50")
plt.show()
```
